```python
from transformers import pipeline

# 1. Define the path where your model was saved
# (Use your Google Drive path if you saved it there, e.g., "/content/drive/My Drive/best_distilbert_sentiment_model")
model_path = "/content/drive/MyDrive/AI/best_distilbert_sentiment_model"

# 2. Create a text-classification pipeline
sentiment_pipeline = pipeline("text-classification", model=model_path)

# 3. Define some new, unseen reviews to test
new_reviews = [
    "This product absolute garbage, stop working immediately!",
    "It is okay, nothing special, average quality for the price.",
    "Amazing quality, fast delivery, I absolutely loved the packaging!",
]

# 4. Run predictions
predictions = sentiment_pipeline(new_reviews)

# 5. Display the results cleanly
for review, pred in zip(new_reviews, predictions):
  print(f"Review: {review}")
  print(
      f"Prediction: {pred['label']} (Confidence: {pred['score']:.4f})\n"
      
  )
```
